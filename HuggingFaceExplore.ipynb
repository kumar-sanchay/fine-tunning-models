```python
from transformers import pipeline
```

```python
import pandas as pd

classifier = pipeline('text-classification', device=0, model='bhadresh-savani/bert-base-uncased-emotion')

```

```python
text = 'I kind of love this not fully'
```

```python
outputs = classifier(text)
```
