```python
from transformers import pipeline
```

```python
import pandas as pd

classifier = pipeline('text-classification', device=0, model='bhadresh-savani/bert-base-uncased-emotion')

```

```python
text = 'I kind of love this not fully'
```

```python
outputs = classifier(text)
```

```python
from transformers import pipeline

# ner = pipeline(task='ner', model='dbmdz/bert-large-cased-finetuned-conll03-english')
ner = pipeline(task='ner', model='vblagoje/bert-english-uncased-finetuned-pos', device=0)
```

```python
text = 'I like Talkies from india'

output = ner(text)
```

```python
import pandas as pd

pd.DataFrame(output)
```

```python
ques_ans = pipeline('table-question-answering', device=0)
```

```python
context = "Steve Jobs founded Apple INC and later it became one of the top company in tech specially with iphone."
question = "Who founded Apple?"

output = ques_ans(question=question, context=context)
output
```
